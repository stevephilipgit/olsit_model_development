#### Date Features

The purchase timestamp contains a lot of information, but it is stored as one datetime value.

Example:

2018-08-08 08:38:49

A business analyst rarely analyses full timestamps.

Instead, we extract useful components such as:

Year
Month
Quarter
Day
Weekday
Hour

These dimensions allow us to analyse sales trends over time.


```python

fact_orders["purchase_year"] = (
    fact_orders["order_purchase_timestamp"].dt.year
)

fact_orders["purchase_month"] = (
    fact_orders["order_purchase_timestamp"].dt.month
)

fact_orders["purchase_month_name"] = (
    fact_orders["order_purchase_timestamp"].dt.month_name()
)

fact_orders["purchase_quarter"] = (
    fact_orders["order_purchase_timestamp"].dt.quarter
)

fact_orders["purchase_day"] = (
    fact_orders["order_purchase_timestamp"].dt.day
)

fact_orders["purchase_weekday"] = (
    fact_orders["order_purchase_timestamp"].dt.day_name()
)

fact_orders["purchase_hour"] = (
    fact_orders["order_purchase_timestamp"].dt.hour
)
```

### Delivery Features (Most Important)

The next section is where we'll calculate business KPIs such as:

Approval Time (hours)

Shipping Time (days)

Delivery Time (days)

Total Order Cycle (days)

Delivery Delay (days)

| Feature                   | Formula               | Business Meaning                 |
| ------------------------- | --------------------- | -------------------------------- |
| `approval_time_hours`     | Approved − Purchased  | Time taken to approve an order   |
| `shipping_time_days`      | Carrier − Approved    | Time taken to dispatch the order |
| `delivery_time_days`      | Delivered − Carrier   | Shipping duration                |
| `total_order_cycle_days`  | Delivered − Purchased | Total customer waiting time      |
| `estimated_delivery_days` | Estimated − Purchased | Promised delivery time           |
| `delivery_delay_days`     | Delivered − Estimated | Early/Late delivery              |


```python
#6.1 Approval Time hours

#Measures how quickly the platform approves an order after it is placed.

fact_orders["approval_time_hours"] = (
    (
        fact_orders["order_approved_at"]
        - fact_orders["order_purchase_timestamp"]
    ).dt.total_seconds() / 3600
)

print("Approval Time Hours - NA Count:", fact_orders["approval_time_hours"].head())
```

```python
# 3.2 Shipping Time

#Time between approval and handing the package to the shipping carrier.


fact_orders["shipping_time_days"] = (
    (
        fact_orders["order_delivered_carrier_date"]
        - fact_orders["order_approved_at"]
    ).dt.days
)
```

```python
#Delivery Time

#Measures the time taken by the carrier to deliver the package after receiving it.

fact_orders["delivery_time_days"] = (

    (

        fact_orders["order_delivered_customer_date"]

        - fact_orders["order_delivered_carrier_date"]

    ).dt.days

)
```

```python
fact_orders.head()
```

```python
#3.4 Total Order Cycle

#Measures the customer's total waiting time. From purchase until delivery.

fact_orders["total_order_cycle_days"] = (
    (
        fact_orders["order_delivered_customer_date"]
        - fact_orders["order_purchase_timestamp"]
    ).dt.days
)
```

```python
#3.5 Estimated Delivery Time

#How many days the company promised for delivery.

fact_orders["estimated_delivery_days"] = (
    (
        fact_orders["order_estimated_delivery_date"]
        - fact_orders["order_purchase_timestamp"]
    ).dt.days
)
```

```python
fact_orders.head()
```

```python
#3.6 Delivery Delay

#Compares the actual delivery date with the estimated delivery date.

fact_orders["delivery_delay_days"] = (
    (
        fact_orders["order_delivered_customer_date"]
        - fact_orders["order_estimated_delivery_date"]
    ).dt.days
)
```

### 7. Sales Fact Features

Unlike fact_orders, sales_fact is at the product level (1 row = 1 product sold). Here we'll create features related to revenue, freight, and product characteristics.

7.1 Total Item Value


The customer pays:

Product Price
Shipping (Freight)

The total value of one item is:

Price + Freight

```python

sales_fact["total_item_value"] = (
    sales_fact["price"] +
    sales_fact["freight_value"]
)
```

### 7.2 Freight Percentage
### Measures how much of the item's value comes from shipping.

```python
sales_fact["freight_percentage"] = (
    sales_fact["freight_value"] /
    sales_fact["price"]
) * 100
```

### 7.3 Heavy Product Flag


Products above the dataset's median weight are marked as heavy.
Using the median is better than choosing an arbitrary threshold because it adapts to the dataset.


```python
median_weight = sales_fact["product_weight_g"].median()

sales_fact["is_heavy_product"] = (
    sales_fact["product_weight_g"] >
    median_weight
)
```

### 7.4 Large Product Flag

We'll estimate a product's package volume.

Formula:

Length × Width × Height

```python


sales_fact["product_volume_cm3"] = (
    sales_fact["product_length_cm"] *
    sales_fact["product_width_cm"] *
    sales_fact["product_height_cm"]
)
```

```python
median_volume = sales_fact["product_volume_cm3"].median()

sales_fact["is_large_product"] = (
    sales_fact["product_volume_cm3"] >
    median_volume
)
```

### 7.5 High Freight Cost Flag

Some products have unusually high shipping costs.

We'll identify them using the 75th percentile (top 25% of freight costs).

```python

freight_threshold = sales_fact["freight_value"].quantile(0.75)

sales_fact["high_freight"] = (
    sales_fact["freight_value"] >
    freight_threshold
)
```
